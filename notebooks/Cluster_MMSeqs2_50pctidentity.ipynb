```python
import subprocess
import os

input_fasta = "/mnt/c/Users/music/OneDrive/Desktop/CompLab/DesignFiles/FASTA/SynADs_syntheticsonly.fasta"
output_prefix = "/mnt/c/Users/music/Downloads/output/cluster_output"
tmp_dir = "/mnt/c/Users/music/Downloads/tmp_dir_1"
min_seq_id = "0.5"

os.makedirs(tmp_dir, exist_ok=True)

cmd = [
    "mmseqs", "easy-cluster",
    input_fasta,
    output_prefix,
    tmp_dir,
    "--min-seq-id", min_seq_id,
    "-c", "1.0",
    "--cov-mode", "1"
]

subprocess.run(cmd, check=True)
```
